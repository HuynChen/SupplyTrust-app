```python
# --- Bước 1: Import các thư viện cần thiết ---
import pandas as pd
import matplotlib.pyplot as plt

# --- Bước 2: Thiết lập để vẽ biểu đồ ngay trong notebook ---
%matplotlib inline

print("Đã import thư viện...")

# --- Bước 3: Đọc file CSV và xử lý dữ liệu (Feature Engineering) ---
try:
    df = pd.read_csv('shipment_data.csv')

    # Chuyển đổi các cột ngày tháng
    date_cols = ['ProductionDate', 'WarehouseEntryDate', 'ShippingDate', 'RetailerArrivalDate']
    for col in date_cols:
        df[col] = pd.to_datetime(df[col])

    # Tạo feature thời gian ở kho
    df['timedelta_warehouse'] = df['ShippingDate'] - df['WarehouseEntryDate']
    df['duration_warehouse'] = df['timedelta_warehouse'].dt.total_seconds() / (24 * 60 * 60)

    # Tạo feature thời gian vận chuyển
    df['timedelta_shipping'] = df['RetailerArrivalDate'] - df['ShippingDate']
    df['duration_shipping'] = df['timedelta_shipping'].dt.total_seconds() / (24 * 60 * 60)

    # Tạo nhãn (0 = normal, 1 = abnormal)
    df['is_abnormal'] = df['Label'].apply(lambda x: 0 if x == 'normal' else 1)

    print("Đã tạo feature xong!")
    print(df[['duration_warehouse', 'duration_shipping', 'is_abnormal']].head()) # In 5 dòng đầu để kiểm tra
    print("\n" + "="*50 + "\n")


    # --- Bước 4: Tính toán thống kê ---
    print("Thống kê tổng quan:")
    print(df[['duration_warehouse', 'duration_shipping']].describe())
    print("\nThống kê của nhóm BÌNH THƯỜNG (is_abnormal = 0):")
    print(df[df['is_abnormal'] == 0][['duration_warehouse', 'duration_shipping']].describe())
    print("\nThống kê của nhóm BẤT THƯỜNG (is_abnormal = 1):")
    print(df[df['is_abnormal'] == 1][['duration_warehouse', 'duration_shipping']].describe())
    print("\n" + "="*50 + "\n")


    # --- Bước 5: Vẽ biểu đồ (Đã sửa lỗi typo) ---
    print("Đang vẽ biểu đồ...")

    # Biểu đồ 1: Thời gian ở kho
    plt.figure(figsize=(12, 5))
    plt.hist(df[df['is_abnormal'] == 0]['duration_warehouse'], bins=50, label='Bình thường (Normal)', alpha=0.7)
    plt.hist(df[df['is_abnormal'] == 1]['duration_warehouse'], bins=50, label='Bất thường (Abnormal)', alpha=0.9)
    plt.title('Phân phối thời gian ở kho (duration_warehouse)')
    plt.xlabel('Số ngày ở kho')
    plt.ylabel('Số lượng lô hàng')
    plt.legend()
    plt.show()

    # Biểu đồ 2: Thời gian vận chuyển
    plt.figure(figsize=(12, 5))
    plt.hist(df[df['is_abnormal'] == 0]['duration_shipping'], bins=50, label='Bình thường (Normal)', alpha=0.7)
    plt.hist(df[df['is_abnormal'] == 1]['duration_shipping'], bins=50, label='Bất thường (Abnormal)', alpha=0.9)
    plt.title('Phân phối thời gian vận chuyển (duration_shipping)')
    plt.xlabel('Số ngày vận chuyển')
    plt.ylabel('Số lượng lô hàng')
    plt.legend()
    plt.show()

    print("\nHoàn tất!")

except FileNotFoundError:
    print(f"LỖI: Không tìm thấy file 'shipment_data.csv'.")
except Exception as e:
    print(f"Đã xảy ra lỗi: {e}")
```

```python
from sklearn.ensemble import IsolationForest

# --- 1. Chuẩn bị dữ liệu cho mô hình ---
# Mô hình AI sẽ chỉ "nhìn" vào 2 cột feature mà chúng ta đã tạo
features = ['duration_warehouse', 'duration_shipping']
X = df[features]

print(f"Bắt đầu huấn luyện mô hình trên {len(X)} lô hàng...")

# --- 2. Khởi tạo mô hình ---
# contamination="auto" hoặc một số nhỏ (ví dụ 0.05) 
# để chỉ định % dữ liệu ta nghi ngờ là bất thường.
model = IsolationForest(contamination='auto', random_state=42)

# --- 3. Huấn luyện (fit) và dự đoán (predict) ---
# Với IsolationForest, .fit_predict sẽ huấn luyện và trả về dự đoán
# Kết quả trả về:
#  1  = Dữ liệu bình thường (inlier)
# -1  = Dữ liệu bất thường (outlier/anomaly)
predictions = model.fit_predict(X)

# --- 4. Lưu kết quả dự đoán của AI vào DataFrame ---
df['ai_prediction'] = predictions

print("Đã huấn luyện xong!")

# --- 5. Xem kết quả ---
# In ra các lô hàng mà AI cho là BẤT THƯỜNG (predicted -1)
print("\nCác lô hàng AI phát hiện là BẤT THƯỜNG:")

# So sánh nhãn 'Label' (sự thật) với 'ai_prediction' (dự đoán của AI)
cols_to_show = ['ShipmentID', 'duration_warehouse', 'duration_shipping', 'Label', 'ai_prediction']
print(df[df['ai_prediction'] == -1][cols_to_show])
```
